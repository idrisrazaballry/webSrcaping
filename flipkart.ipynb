```python
from selenium import webdriver
import time
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

driver=webdriver.Chrome()

url="https://www.flipkart.com/"
driver.get(url)
time.sleep(3)

driver.find_element(By.CLASS_NAME,'b3wTlE').click()

search_box=driver.find_element(By.NAME,'q')
search_box.send_keys("samsung mobiles")
search_box.send_keys(Keys.ENTER)
time.sleep(3)

# mb_name=driver.find_element(By.CLASS_NAME,'RG5Slk').text

div_list=driver.find_elements(By.CLASS_NAME,'jIjQ8S')
mobile_names,price=[],[]
for d in div_list:
    mobile_names.append(d.find_element(By.CLASS_NAME,'RG5Slk').text)
    try:
        price.append(d.find_element(By.CSS_SELECTOR,'.hZ3P6w.DeU9vF').text)
    except:
        price.append(None)

time.sleep(3)
driver.quit()

# print(mb_name)
print(mobile_names)
print(len(mobile_names))
```

```python
from selenium import webdriver
import time
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

driver=webdriver.Chrome()

url="https://www.flipkart.com/"
driver.get(url)
time.sleep(3)

driver.find_element(By.CLASS_NAME,'b3wTlE').click()

search_box=driver.find_element(By.NAME,'q')
search_box.send_keys("samsung mobiles")
search_box.send_keys(Keys.ENTER)
time.sleep(3)

# mb_name=driver.find_element(By.CLASS_NAME,'RG5Slk').text


mobile_names,price=[],[]
for i in range(3):
    div_list=driver.find_elements(By.CLASS_NAME,'jIjQ8S')
    for d in div_list:
        mobile_names.append(d.find_element(By.CLASS_NAME,'RG5Slk').text)
        try:
            price.append(d.find_element(By.CSS_SELECTOR,'.hZ3P6w.DeU9vF').text)
        except:
            price.append(None)


    driver.find_elements(By.CLASS_NAME,'jgg0SZ')[-1].click()
    time.sleep(3)



time.sleep(3)
driver.quit()
#---------------------------------------------------------------------------#


# print(mb_name)
print(mobile_names,price,sep='\n\n')
print(len(mobile_names))
print(len(price))
```

```python
d={'mobile_names':mobile_names,'cost':price}

import pandas as pd

df=pd.DataFrame(d)
df.to_excel('smasung_mobile.xlsx',index=False)
```
