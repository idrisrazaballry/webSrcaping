extract rating and availability

### Extracting the book details from all 50 pages

```python
names,prices,ratings,stock=[],[],[],[]

for i in range(1,51):
    url=f"https://books.toscrape.com/catalogue/page-{i}.html"
    # print(url)
    
    response=requests.get(url)
    code=response.text
    soup=BeautifulSoup(code)
    
    art_lst=soup.find_all('article')
    
    for art in art_lst:
        names.append(art.find('h3').find('a').attrs['title'])
        prices.append(art.find_all('div')[-1].find('p').text)
        ratings.append(art.find('p').attrs['class'][1])
        stock.append(art.find_all('div')[-1].find_all('p')[1].text.strip())
```

```python
print(names)
print(prices)
print(ratings)
print(stock)
```

```python
import pandas as pd
```

```python
d={'book_names':names,'prices':prices,'ratings':ratings,'availability':stock}
df=pd.DataFrame(d)
df.to_excel('books.xlsx',index=False)
```
