**Web page**
A single document on the web, usually written in HTML, that a browser can display. It's one "screen" of content — text, images, links, forms. Example: `example.com/about.html` is one page.

**Website**
A collection of related web pages hosted under one domain and linked together. Example: `wikipedia.org` is the website; the article on Bengaluru is a page within it.

**Client and server**
- **Client** — the machine/program that asks for something. Your browser, or your Python script running `requests`, is a client.
- **Server** — the machine/program that stores resources and answers those asks. It listens for incoming requests and sends back data.

The relationship is one-directional in initiation: the client always starts the conversation, the server responds.

**Request and response**
- **Request** — what the client sends. It carries a method (`GET` to fetch, `POST` to submit), a URL, and headers (like `User-Agent`, cookies). Optionally a body.
- **Response** — what the server sends back. It carries a status code (200 = OK, 404 = not found, 403 = forbidden, 500 = server error), headers, and the body (the HTML, JSON, image, etc.).

In Python:
```python
r = requests.get("https://example.com")   # the request
print(r.status_code)                      # part of the response
print(r.text)                             # the response body — HTML
```

**Web scraping**
Automatically extracting data from web pages using a program instead of copying it by hand. The program sends requests, receives HTML, and pulls out the specific values it needs.

The basics you need to do it:
1. **HTML/CSS knowledge** — to know what you're looking at and how to target elements (tags, classes, IDs).
2. **HTTP basics** — requests, responses, status codes, headers.
3. **A fetching tool** — `requests` for plain HTTP, or Selenium when JavaScript must run.
4. **A parser** — BeautifulSoup (or lxml) to navigate the HTML tree.
5. **Selectors** — CSS selectors or XPath to locate the data.
6. **Storage** — write results to CSV, JSON, or a database (`csv`, `json`, `pandas`).
7. **Ethics and legality** — check `robots.txt` and the site's terms, rate-limit your requests, don't scrape personal data or hammer a server.

**Static vs dynamic web pages**

| | Static | Dynamic |
|---|---|---|
| Content | Fixed; same HTML for everyone | Generated per request/user |
| Built by | Pre-written HTML files | Server code (PHP, Django) or JavaScript in the browser |
| Changes | Only when someone edits the file | Changes with data, user, time |
| Example | A personal portfolio page | Amazon search results, a Twitter feed |
| Scraping | `requests` + BeautifulSoup is enough | Often needs Selenium, or you find the underlying API the page calls |

This distinction is the single most practical one for scraping. If `r.text` doesn't contain the data you see in the browser, the page is rendering it with JavaScript — that's when you switch to Selenium.

**10 important HTML tags**

1. `<html>` — root element wrapping the whole document
2. `<head>` — metadata: title, links to CSS, scripts
3. `<body>` — everything visible on the page
4. `<h1>`–`<h6>` — headings, in descending importance
5. `<p>` — paragraph of text
6. `<a>` — anchor/hyperlink; the `href` attribute holds the URL
7. `<div>` — generic block container, used constantly for layout and grouping
8. `<span>` — generic inline container, wraps text inside a line
9. `<table>` (with `<tr>`, `<th>`, `<td>`) — tabular data; a scraper's favourite target
10. `<img>` — image; the `src` attribute holds the file URL

Worth adding: `<ul>`/`<ol>`/`<li>` for lists and `<form>`/`<input>` for user input — both come up a lot in scraping.

For scraping specifically, the tags matter less than the **attributes** on them — `class`, `id`, and `href` are what you'll actually use to target elements:

```python
soup.find_all("a", class_="product-link")
soup.select("div#main table tr")
```

````python
import requests
from bs4 import BeautifulSoup # extract the webscraping content
````

````python
url = "https://books.toscrape.com/"

requests.get(url) # requesting and respond both task is done by get() method from server using given url
````

````python
response = requests.get(url)
code = response.text
````

````python
soup = BeautifulSoup(code)
soup
````

````python
type(soup)
````

### Extracting the first book name

````python
soup.find("article") # find method gives the first occurance of title tag(eg: article)
````

````python
book_name = soup.find("article").find('h3').find('a').text
print(book_name)
````

````python
soup.find("article").find('h3').find('a').attrs
````

````python
full_book_name = soup.find("article").find('h3').find('a').attrs['title']
print(full_book_name)
````
