```python
from selenium import webdriver
import time
from selenium.webdriver.common.by import By


driver = webdriver.Chrome()

url = "https://www.flipkart.com/"
driver.get(url)
time.sleep(3)

driver.find_element(By.CLASS_NAME, 'b3wTlE').click()

search_box = driver.find_element(By)


div_list = driver.find_elements(By.CLASS_NAME, 'jIjQ8S')
mobile_names, prices = [], []
for d in div_list:
    mobile_names.append(d.find_element(By.CLASS_NAME, 'RG5Slk').text)
    try:
        prices.append(d.find_element(By.CSS_SELECTOR, '.hZ3P6w.DeU9vF').text)
    except:
        prices.append(None)

time.sleep(3)
driver.quit()

print(mobile_names)


print(mobile_names, prices, end='\n\n')
print(len(mobile_names))
print(len(prices))
```
